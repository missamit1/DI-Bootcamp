## Exercise 1 — Open Source Levels Reflection

### Fully Open

- **What is open:** architecture, weights, training/inference code and sufficient training-data information, under terms permitting use, study, modification and sharing.
- **What this enables:** inspect and adapt the pipeline or retrain it; full reproduction still requires data access, compute and a complete recipe.

### Weights Released

- **What is open:** trained parameters, usually with an architecture definition and inference implementation; the complete training pipeline or data may be unavailable.
- **What this enables:** run locally and fine-tune when the license permits; cannot assume the original training process is reproducible or that commercial use is allowed.

### Architecture Only

- **What is open:** the model design, described in a paper or code, without its pretrained parameters.
- **What this enables:** inspect and implement the structure; using the original learned behavior requires unavailable weights, so a new model must first be trained.

| Level | Available material | Impact on adaptation |
|---|---|---|
| Fully Open | Design, parameters, code and data information | Broadest ability to inspect and change the training process |
| Weights Released | Pretrained parameters and supporting implementation | Fine-tuning is possible if the license allows it |
| Architecture Only | Structural specification | No pretrained checkpoint to adapt directly |

**Comparative paragraph (4 sentences).** Fully open releases expose the components needed to understand and modify a model's development process. Weight releases make practical adaptation possible but may leave important questions about training unanswered. Architecture-only releases explain the design without supplying the knowledge encoded in trained parameters. These levels describe access, while the license determines which uses are permitted.

**Healthcare answer (2 sentences).** Accessible pretrained weights, compatible training code and permission to fine-tune are essential for adapting an existing assistant to clinical data; a fully open release is preferable for auditability but is not technically required. Neither access level guarantees privacy or clinical reliability, so adaptation should use appropriately governed data and a separate clinical evaluation.

The classroom categories are a simplification. The [Open Source AI Definition 1.0](https://opensource.org/ai/open-source-ai-definition) requires data information, code and parameters; it does not require unrestricted publication of every private training record.

### Approximate weight storage

The calculation below is a lower bound for packed 4-bit weights in decimal GB. GGUF metadata, mixed-precision tensors, quantization scales, KV cache, activations and runtime overhead add to it; it is not a measured file size or total RAM requirement.


```python
parameters_billions = {'Qwen 1.5B': 1.54, 'Qwen 3B': 3.09, 'Phi 3.5 mini': 3.8, 'Generic 7B': 7.0}
for name, parameters in parameters_billions.items():
    weight_gb = parameters * 4 / 8
    print(f'{name}: {weight_gb:.2f} GB of idealized 4-bit weights')
```

## Exercise 4 — Local Readiness Audit

The following cell reads the machine on which it runs. Its saved output was collected on the local Mac on 29 October 2026; rerunning it in Colab describes Colab's VM instead. It reads memory, disk and compiler information without modifying the system.


```python
import platform
import shutil
import subprocess
from pathlib import Path
from datetime import datetime, timezone

def command_output(arguments):
    try:
        result = subprocess.run(arguments, capture_output=True, text=True, timeout=10, check=True)
        return result.stdout.strip()
    except (OSError, subprocess.SubprocessError):
        return 'Unavailable'

system = platform.system()
if system == 'Darwin':
    memory_text = command_output(['/usr/sbin/sysctl', '-n', 'hw.memsize'])
    ram_gib = int(memory_text) / 1024 ** 3 if memory_text.isdigit() else None
    processor = command_output(['/usr/sbin/sysctl', '-n', 'machdep.cpu.brand_string'])
    os_version = 'macOS ' + platform.mac_ver()[0]
elif system == 'Linux':
    memory_lines = Path('/proc/meminfo').read_text().splitlines()
    memory_kib = int(next(line for line in memory_lines if line.startswith('MemTotal:')).split()[1])
    ram_gib = memory_kib / 1024 ** 2
    processor = platform.processor() or platform.machine()
    os_version = 'Linux ' + platform.release()
else:
    ram_gib = None
    processor = platform.processor() or platform.machine()
    os_version = platform.platform()

free_gib = shutil.disk_usage(Path.cwd()).free / 1024 ** 3
print('Observed:', datetime.now(timezone.utc).strftime('%Y-%m-%d %H:%M UTC'))
print('CPU:', processor, '/', platform.machine())
print('OS:', os_version)
print('RAM:', f'{ram_gib:.1f} GiB' if ram_gib is not None else 'Check manually')
print('Free disk:', f'{free_gib:.1f} GiB')
print('CMake on PATH:', 'Yes' if shutil.which('cmake') else 'No')
compiler = shutil.which('clang') or shutil.which('gcc')
print('Compiler:', command_output([compiler, '--version']).splitlines()[0] if compiler else 'Not found')
```

## Exercise 6 — Cloud vs. Local Deployment Plan

### Local Deployment / Cloud Deployment

- **Cost:** local use can reuse the existing machine, but electricity, maintenance and eventual replacement still cost money; cloud avoids a hardware purchase but charges for allocated resources, storage and potentially network traffic.
- **Latency:** local inference avoids internet round trips but may generate slowly on limited hardware; cloud GPUs may generate faster, while network delay, queuing and cold starts add latency.
- **Scalability:** a local machine offers predictable personal capacity but is limited by its memory and availability; cloud can add capacity, subject to quotas, regional availability and spending limits.
- **Security and control:** local processing can keep prompts on the device if the software is configured accordingly, but the owner must secure it; cloud requires suitable access controls, data-handling terms and region choices while offering managed security features.
- **Maintenance:** local deployment provides direct control over versions but requires setup, updates and backups; managed cloud reduces some operational work but adds provider dependence and still requires application monitoring.

**Plan:** start with the existing Mac and a small quantized model for private experimentation. Keep a cloud option for larger models or short bursts, using synthetic/non-sensitive prompts during evaluation. Compare both environments on an identical workload before choosing production hosting.

### Illustrative cost calculation

Assume 40 compute-hours per month, an incremental local load of 0.08 kW and electricity at $0.25/kWh. Compare that with a hypothetical $0.50/hour cloud machine plus $5/month storage. These are **chosen scenario inputs, not measured power draw, an electricity tariff quote or a current provider offer**. The calculation excludes administration, taxes, depreciation and data transfer; for a new machine, add its amortized purchase cost.


```python
hours_per_month = 40
incremental_local_kw = 0.08
electricity_usd_per_kwh = 0.25
cloud_usd_per_hour = 0.50
cloud_storage_usd = 5
local_purchase_usd = 0  # Existing computer; replace this for a new purchase.
amortization_months = 36

local_monthly = hours_per_month * incremental_local_kw * electricity_usd_per_kwh
local_monthly += local_purchase_usd / amortization_months
cloud_monthly = hours_per_month * cloud_usd_per_hour + cloud_storage_usd
print(f'Illustrative local monthly cost: ${local_monthly:.2f}')
print(f'Illustrative cloud monthly cost: ${cloud_monthly:.2f}')
print('Compare actual throughput and total operating costs before treating these as equivalent services.')
```
